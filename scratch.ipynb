```python
# Standard Packages

# Snowflake Packages
from snowflake.snowpark.context import get_active_session
from snowflake.snowpark.types import *
from snowflake.snowpark.types import StructType, StructField, StringType, IntegerType

# Third Party
import polars as pl
# Modules
import main
```

```python
#df:pl.DataFrame = main.SNHD_INSPECTIONS()
```

```python
session = get_active_session()
staged_file = '@PC_DBT_DB.PUBLIC.SNHD/Restaurant_Inspections_Open_Data_-5146138254430322373.csv'

schema = StructType([
    StructField('Serial_Number',StringType()),
    StructField('Permit_Number',StringType()),
    StructField('Restaurant_Name',StringType()),
    StructField('Location_Name',StringType()),
    StructField('Category_Name',StringType()),
    StructField('Address',StringType()),
    StructField('City',StringType()),
    StructField('State',StringType()),
    StructField('Zip',StringType()),
    StructField('Current_Demerits',StringType()),
    StructField('Current_Grade',StringType()),
    StructField('Date_Current',StringType()),
    StructField('Inspection_Date',StringType()),
    StructField('Inspection_Time',StringType()),
    StructField('Employee_ID',StringType()),
    StructField('Inspection_Type',StringType()),
    StructField('Inspection_Demerits',IntegerType()),
    StructField('Inspection_Grade',StringType()),
    StructField('Permit_Status',StringType()),
    StructField('Inspection_Result',StringType()),
    StructField('Violations',StringType()),
    StructField('Record_Updated',StringType()),
    StructField('Location_1',StringType()),
    StructField('ObjectId',IntegerType()),
])
# Read the CSV from the stage into a Snowpark DataFrame
df = session.read.options({
    "field_delimiter": ",",
    "skip_header": 1,          # skip the header row
    "infer_schema": True,       # auto-detect column types
    #"parse_header": True,
}).csv(staged_file)

df = session.read.schema(schema).options({
    "field_delimiter": ",",
    "skip_header": 1
}).csv(staged_file)

# Write it to a table (creates the table if it doesn't exist)
df.write.mode("overwrite").save_as_table("PC_DBT_DB.PUBLIC.SNHD")


```
